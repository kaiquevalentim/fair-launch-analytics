## Histogram analysis

```python
# Histogram of LBPs participants, txns, volume_usd

columns = ['participants', 'txns', 'volume_usd']

# Add a gap between subplots
fig, axs = plt.subplots(2, 3, figsize=(18, 10))
plt.subplots_adjust(hspace=0.4, wspace=0.4)

for i, df in enumerate(list_of_lbps[:-1]):
  for j, col in enumerate(columns):
      axs[i][j].hist(df[col], bins=30, alpha=0.5, label=f'V{i+1} LBPs')
      axs[i][j].set_title(f'Version {i+1} LBP ' + col.capitalize() + ' Distribution')
      axs[i][j].set_xlabel('Number of ' + col)
      axs[i][j].set_ylabel('Frequency')
      axs[i][j].legend()
```
